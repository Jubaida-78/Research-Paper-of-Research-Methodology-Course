```python
!pip -q install xgboost shap openpyxl

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import permutation_importance

from xgboost import XGBClassifier
import shap

print("Libraries loaded successfully.")

BASE = "/content"

BD_PATH = os.path.join(BASE, "BD-KDD Dataset.csv")
CKD_PATH = os.path.join(BASE, "CKD_dataset.csv")

print("BD-KDD exists:", os.path.exists(BD_PATH))
print("CKD exists:", os.path.exists(CKD_PATH))

if not os.path.exists(BD_PATH):
    raise FileNotFoundError("BD-KDD Dataset.csv was not found in /content")

if not os.path.exists(CKD_PATH):
    raise FileNotFoundError("CKD_dataset.csv was not found in /content")

bd = pd.read_csv(BD_PATH)
ckd = pd.read_csv(CKD_PATH)

print("\nBD-KDD shape:", bd.shape)
print("CKD shape:", ckd.shape)

print("\nBD-KDD columns:")
print(bd.columns.tolist())

print("\nCKD columns:")
print(ckd.columns.tolist())


def normalize_column_name(c):
    c = str(c).strip().lower()
    c = c.replace(" ", "_")
    c = c.replace("-", "_")
    c = c.replace("/", "_")
    c = c.replace(".", "_")
    c = c.replace("(", "")
    c = c.replace(")", "")
    return c


bd.columns = [normalize_column_name(c) for c in bd.columns]
ckd.columns = [normalize_column_name(c) for c in ckd.columns]

print("\nNormalized BD-KDD columns:")
print(bd.columns.tolist())

print("\nNormalized CKD columns:")
print(ckd.columns.tolist())


def find_target(df):
    candidates = [
        "class",
        "classification",
        "target",
        "label",
        "diagnosis"
    ]

    for c in candidates:
        if c in df.columns:
            return c

    raise ValueError("Target column could not be identified.")


bd_target = find_target(bd)
ckd_target = find_target(ckd)

print("\nBD target:", bd_target)
print("CKD target:", ckd_target)


def convert_target(series):
    s = series.copy()

    if pd.api.types.is_numeric_dtype(s):
        vals = sorted(pd.Series(s).dropna().unique().tolist())

        if set(vals).issubset({0, 1}):
            return s.astype(int)

        if len(vals) == 2:
            mapping = {vals[0]: 0, vals[1]: 1}
            return s.map(mapping).astype(int)

    s2 = s.astype(str).str.strip().str.lower()

    positive = {
        "1",
        "yes",
        "y",
        "ckd",
        "disease",
        "kidney disease",
        "positive",
        "present",
        "true"
    }

    negative = {
        "0",
        "no",
        "n",
        "notckd",
        "not ckd",
        "healthy",
        "negative",
        "absent",
        "false"
    }

    result = []

    for v in s2:
        if v in positive:
            result.append(1)
        elif v in negative:
            result.append(0)
        else:
            result.append(np.nan)

    return pd.Series(result, index=series.index)


bd["target_binary"] = convert_target(bd[bd_target])
ckd["target_binary"] = convert_target(ckd[ckd_target])

print("\nBD target distribution:")
print(bd["target_binary"].value_counts(dropna=False))

print("\nCKD target distribution:")
print(ckd["target_binary"].value_counts(dropna=False))


common_feature_mapping = {
    "age": {
        "ckd": "age",
        "bd": "age"
    },
    "haemoglobin": {
        "ckd": "haemoglobin",
        "bd": "hemo"
    },
    "hematocrit": {
        "ckd": "hematocrit",
        "bd": "pcv"
    },
    "rbc_count": {
        "ckd": "rbc_count",
        "bd": "rbcc"
    },
    "wbc_count": {
        "ckd": "wbc_count",
        "bd": "wbcc"
    },
    "creatinine": {
        "ckd": "creatinine",
        "bd": "sc"
    },
    "urea": {
        "ckd": "urea",
        "bd": "bu"
    }
}

print("\nChecking harmonized features:")

for common_name, mapping in common_feature_mapping.items():
    print(
        common_name,
        "-> CKD:",
        mapping["ckd"],
        "| BD-KDD:",
        mapping["bd"]
    )

    if mapping["ckd"] not in ckd.columns:
        raise ValueError(
            f"CKD feature missing: {mapping['ckd']}"
        )

    if mapping["bd"] not in bd.columns:
        raise ValueError(
            f"BD-KDD feature missing: {mapping['bd']}"
        )


feature_names = list(common_feature_mapping.keys())

X_ckd = pd.DataFrame()

for common_name, mapping in common_feature_mapping.items():
    X_ckd[common_name] = pd.to_numeric(
        ckd[mapping["ckd"]],
        errors="coerce"
    )

X_bd = pd.DataFrame()

for common_name, mapping in common_feature_mapping.items():
    X_bd[common_name] = pd.to_numeric(
        bd[mapping["bd"]],
        errors="coerce"
    )

y_ckd = ckd["target_binary"]
y_bd = bd["target_binary"]


valid_ckd = y_ckd.notna()
valid_bd = y_bd.notna()

X_ckd = X_ckd.loc[valid_ckd].reset_index(drop=True)
y_ckd = y_ckd.loc[valid_ckd].astype(int).reset_index(drop=True)

X_bd = X_bd.loc[valid_bd].reset_index(drop=True)
y_bd = y_bd.loc[valid_bd].astype(int).reset_index(drop=True)

print("\nFinal CKD analysis shape:", X_ckd.shape)
print("Final BD-KDD analysis shape:", X_bd.shape)

print("\nFeatures used:")
print(feature_names)

print("\nCKD missing values:")
print(X_ckd.isna().sum())

print("\nBD-KDD missing values:")
print(X_bd.isna().sum())

print("\nCKD classes:")
print(y_ckd.value_counts())

print("\nBD-KDD classes:")
print(y_bd.value_counts())


X_train, X_test, y_train, y_test = train_test_split(
    X_ckd,
    y_ckd,
    test_size=0.20,
    stratify=y_ckd,
    random_state=42
)

print("\nInternal CKD training set:", X_train.shape)
print("Internal CKD test set:", X_test.shape)


models = {
    "Logistic Regression": Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            LogisticRegression(
                max_iter=3000,
                class_weight="balanced",
                random_state=42
            )
        )
    ]),

    "Random Forest": Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "model",
            RandomForestClassifier(
                n_estimators=500,
                max_depth=None,
                min_samples_split=2,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ]),

    "XGBoost": Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "model",
            XGBClassifier(
                n_estimators=400,
                max_depth=4,
                learning_rate=0.03,
                subsample=0.9,
                colsample_bytree=0.9,
                objective="binary:logistic",
                eval_metric="logloss",
                random_state=42,
                n_jobs=-1
            )
        )
    ])
}


def evaluate_model(model, X, y):
    pred = model.predict(X)

    if hasattr(model, "predict_proba"):
        prob = model.predict_proba(X)[:, 1]
    else:
        prob = None

    result = {
        "Accuracy": accuracy_score(y, pred),
        "Precision": precision_score(y, pred, zero_division=0),
        "Recall": recall_score(y, pred, zero_division=0),
        "F1": f1_score(y, pred, zero_division=0)
    }

    if prob is not None and len(np.unique(y)) == 2:
        result["ROC_AUC"] = roc_auc_score(y, prob)
    else:
        result["ROC_AUC"] = np.nan

    return result, pred, prob


internal_results = []
external_results = []

trained_models = {}


for name, model in models.items():

    print("\n" + "=" * 70)
    print("TRAINING:", name)
    print("=" * 70)

    model.fit(X_train, y_train)

    trained_models[name] = model

    internal_metrics, internal_pred, internal_prob = evaluate_model(
        model,
        X_test,
        y_test
    )

    external_metrics, external_pred, external_prob = evaluate_model(
        model,
        X_bd,
        y_bd
    )

    internal_row = {
        "Model": name,
        **internal_metrics
    }

    external_row = {
        "Model": name,
        **external_metrics
    }

    internal_results.append(internal_row)
    external_results.append(external_row)

    print("\nInternal CKD performance:")
    for k, v in internal_metrics.items():
        print(f"{k}: {v:.4f}")

    print("\nExternal BD-KDD performance:")
    for k, v in external_metrics.items():
        print(f"{k}: {v:.4f}")


internal_results_df = pd.DataFrame(internal_results)
external_results_df = pd.DataFrame(external_results)

print("\n\nINTERNAL CKD RESULTS")
display(internal_results_df.round(4))

print("\nEXTERNAL BD-KDD RESULTS")
display(external_results_df.round(4))


comparison_df = internal_results_df.merge(
    external_results_df,
    on="Model",
    suffixes=("_Internal_CKD", "_External_BD_KDD")
)

comparison_df["Accuracy_Gap"] = (
    comparison_df["Accuracy_Internal_CKD"]
    - comparison_df["Accuracy_External_BD_KDD"]
)

comparison_df["F1_Gap"] = (
    comparison_df["F1_Internal_CKD"]
    - comparison_df["F1_External_BD_KDD"]
)

comparison_df["AUC_Gap"] = (
    comparison_df["ROC_AUC_Internal_CKD"]
    - comparison_df["ROC_AUC_External_BD_KDD"]
)

print("\n\nCROSS-POPULATION PERFORMANCE COMPARISON")
display(comparison_df.round(4))


best_model_name = (
    external_results_df
    .sort_values("ROC_AUC", ascending=False)
    .iloc[0]["Model"]
)

best_model = trained_models[best_model_name]

print("\nBEST EXTERNAL MODEL:")
print(best_model_name)


external_pred = best_model.predict(X_bd)
external_prob = best_model.predict_proba(X_bd)[:, 1]

print("\nExternal classification report:")
print(
    classification_report(
        y_bd,
        external_pred,
        target_names=["Healthy", "Kidney Disease"],
        digits=4
    )
)


cm = confusion_matrix(y_bd, external_pred)

plt.figure(figsize=(6, 5))
plt.imshow(cm)
plt.title(f"External Validation Confusion Matrix - {best_model_name}")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks([0, 1], ["Healthy", "Kidney Disease"])
plt.yticks([0, 1], ["Healthy", "Kidney Disease"])

for i in range(2):
    for j in range(2):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )

plt.colorbar()
plt.tight_layout()
plt.show()


fpr_internal, tpr_internal, _ = roc_curve(
    y_test,
    trained_models[best_model_name].predict_proba(X_test)[:, 1]
)

fpr_external, tpr_external, _ = roc_curve(
    y_bd,
    external_prob
)

auc_internal = roc_auc_score(
    y_test,
    trained_models[best_model_name].predict_proba(X_test)[:, 1]
)

auc_external = roc_auc_score(
    y_bd,
    external_prob
)

plt.figure(figsize=(7, 6))

plt.plot(
    fpr_internal,
    tpr_internal,
    label=f"Internal CKD AUC = {auc_internal:.3f}"
)

plt.plot(
    fpr_external,
    tpr_external,
    label=f"External BD-KDD AUC = {auc_external:.3f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title(
    f"Internal vs External ROC - {best_model_name}"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


perm = permutation_importance(
    best_model,
    X_bd,
    y_bd,
    n_repeats=30,
    random_state=42,
    scoring="roc_auc",
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "Feature": X_bd.columns,
    "Importance_Mean": perm.importances_mean,
    "Importance_SD": perm.importances_std
})

importance_df = (
    importance_df
    .sort_values(
        "Importance_Mean",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nEXTERNAL BD-KDD PERMUTATION IMPORTANCE")
display(importance_df.round(4))


plt.figure(figsize=(8, 5))

plt.barh(
    importance_df["Feature"][::-1],
    importance_df["Importance_Mean"][::-1]
)

plt.xlabel("Mean decrease in ROC-AUC")
plt.ylabel("Clinical Feature")
plt.title(
    f"External Feature Importance - {best_model_name}"
)

plt.tight_layout()
plt.show()


try:
    xgb_model = trained_models["XGBoost"]

    xgb_inner = xgb_model.named_steps["model"]

    X_bd_imputed = (
        xgb_model
        .named_steps["imputer"]
        .transform(X_bd)
    )

    shap_explainer = shap.TreeExplainer(xgb_inner)

    shap_values = shap_explainer.shap_values(
        X_bd_imputed
    )

    print("\nSHAP analysis generated successfully.")

    shap.summary_plot(
        shap_values,
        X_bd_imputed,
        feature_names=feature_names,
        show=True
    )

except Exception as e:
    print("\nSHAP analysis could not be generated:")
    print(e)


def bootstrap_auc(y_true, y_prob, n_bootstrap=1000, random_state=42):
    rng = np.random.RandomState(random_state)
    aucs = []

    y_true = np.asarray(y_true)
    y_prob = np.asarray(y_prob)

    for _ in range(n_bootstrap):
        indices = rng.randint(
            0,
            len(y_true),
            len(y_true)
        )

        y_sample = y_true[indices]
        p_sample = y_prob[indices]

        if len(np.unique(y_sample)) < 2:
            continue

        aucs.append(
            roc_auc_score(
                y_sample,
                p_sample
            )
        )

    aucs = np.array(aucs)

    return (
        np.mean(aucs),
        np.percentile(aucs, 2.5),
        np.percentile(aucs, 97.5)
    )


mean_auc, lower_auc, upper_auc = bootstrap_auc(
    y_bd,
    external_prob
)

print("\nExternal BD-KDD bootstrap ROC-AUC:")
print(f"Mean AUC: {mean_auc:.4f}")
print(f"95% CI: [{lower_auc:.4f}, {upper_auc:.4f}]")


feature_importance_table = importance_df.copy()

feature_importance_table["Rank"] = (
    np.arange(1, len(feature_importance_table) + 1)
)

feature_importance_table = feature_importance_table[
    [
        "Rank",
        "Feature",
        "Importance_Mean",
        "Importance_SD"
    ]
]


dataset_summary = pd.DataFrame({
    "Dataset": [
        "CKD",
        "BD-KDD"
    ],
    "Records": [
        len(X_ckd),
        len(X_bd)
    ],
    "Features_Used": [
        X_ckd.shape[1],
        X_bd.shape[1]
    ],
    "Healthy": [
        int((y_ckd == 0).sum()),
        int((y_bd == 0).sum())
    ],
    "Kidney_Disease": [
        int((y_ckd == 1).sum()),
        int((y_bd == 1).sum())
    ]
})


methodology_table = pd.DataFrame({
    "Stage": [
        "Source training population",
        "External validation population",
        "Feature harmonization",
        "Internal split",
        "External evaluation",
        "Explainability",
        "Feature importance",
        "Primary external metric"
    ],
    "Description": [
        "CKD benchmark dataset",
        "Independent Bangladeshi BD-KDD dataset",
        "Seven clinically comparable biomarkers",
        "80% training / 20% internal testing",
        "Entire independent BD-KDD dataset",
        "SHAP",
        "Permutation importance",
        "ROC-AUC"
    ]
})


output_file = "/content/CKD_CrossPopulation_XAI_Results.xlsx"

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    dataset_summary.to_excel(
        writer,
        sheet_name="Dataset_Summary",
        index=False
    )

    methodology_table.to_excel(
        writer,
        sheet_name="Methodology",
        index=False
    )

    internal_results_df.to_excel(
        writer,
        sheet_name="Internal_CKD",
        index=False
    )

    external_results_df.to_excel(
        writer,
        sheet_name="External_BD_KDD",
        index=False
    )

    comparison_df.to_excel(
        writer,
        sheet_name="Cross_Population",
        index=False
    )

    feature_importance_table.to_excel(
        writer,
        sheet_name="Feature_Importance",
        index=False
    )


comparison_df.to_csv(
    "/content/Cross_Population_Model_Comparison.csv",
    index=False
)

importance_df.to_csv(
    "/content/External_Feature_Importance.csv",
    index=False
)

print("\n" + "=" * 70)
print("RESEARCH PIPELINE COMPLETED")
print("=" * 70)

print("\nBest external model:")
print(best_model_name)

print("\nExternal BD-KDD AUC:")
print(f"{auc_external:.4f}")

print("\nExternal BD-KDD F1:")
print(
    f"{f1_score(y_bd, external_pred):.4f}"
)

print("\nExternal BD-KDD Accuracy:")
print(
    f"{accuracy_score(y_bd, external_pred):.4f}"
)

print("\nBootstrap AUC 95% CI:")
print(
    f"[{lower_auc:.4f}, {upper_auc:.4f}]"
)

print("\nResults saved to:")
print(output_file)

print("\nCSV files:")
print("/content/Cross_Population_Model_Comparison.csv")
print("/content/External_Feature_Importance.csv")
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = []

for name, model in models.items():

    scores = cross_validate(
        model,
        X_ckd,
        y_ckd,
        cv=cv,
        scoring=[
            "accuracy",
            "precision",
            "recall",
            "f1",
            "roc_auc"
        ],
        n_jobs=-1
    )

    cv_results.append({
        "Model": name,
        "CV_Accuracy_Mean": scores["test_accuracy"].mean(),
        "CV_Accuracy_SD": scores["test_accuracy"].std(),
        "CV_Precision_Mean": scores["test_precision"].mean(),
        "CV_Recall_Mean": scores["test_recall"].mean(),
        "CV_F1_Mean": scores["test_f1"].mean(),
        "CV_ROC_AUC_Mean": scores["test_roc_auc"].mean(),
        "CV_ROC_AUC_SD": scores["test_roc_auc"].std()
    })

cv_results_df = pd.DataFrame(cv_results)

display(cv_results_df.round(4))
```

```python
best_source_model_name = (
    cv_results_df
    .sort_values("CV_ROC_AUC_Mean", ascending=False)
    .iloc[0]["Model"]
)

print("Best model selected using UCI 5-fold CV:")
print(best_source_model_name)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)
from sklearn.inspection import permutation_importance

# ============================================================
# 1. FIX WBC UNIT ONLY IF NECESSARY
# ============================================================

wbc_mean = X_bd["wbc_count"].mean()

if wbc_mean < 1:
    X_bd["wbc_count"] = X_bd["wbc_count"] * 1000
elif wbc_mean > 100:
    X_bd["wbc_count"] = X_bd["wbc_count"] / 1000

print("CKD WBC mean:", X_ckd["wbc_count"].mean())
print("BD-KDD WBC mean:", X_bd["wbc_count"].mean())

# ============================================================
# 2. INTERNAL TRAIN/TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_ckd,
    y_ckd,
    test_size=0.20,
    stratify=y_ckd,
    random_state=42
)

# ============================================================
# 3. TRAIN ALL THREE MODELS AGAIN
# ============================================================

trained_models = {}

for name, model in models.items():
    print("\nTraining:", name)
    model.fit(X_train, y_train)
    trained_models[name] = model

# ============================================================
# 4. EVALUATION FUNCTION
# ============================================================

def evaluate(model, X, y):
    pred = model.predict(X)
    prob = model.predict_proba(X)[:, 1]

    return {
        "Accuracy": accuracy_score(y, pred),
        "Precision": precision_score(y, pred, zero_division=0),
        "Recall": recall_score(y, pred, zero_division=0),
        "F1": f1_score(y, pred, zero_division=0),
        "ROC_AUC": roc_auc_score(y, prob)
    }

# ============================================================
# 5. INTERNAL + EXTERNAL RESULTS
# ============================================================

internal_results = []
external_results = []

for name, model in trained_models.items():

    internal = evaluate(model, X_test, y_test)
    external = evaluate(model, X_bd, y_bd)

    internal_results.append({
        "Model": name,
        **internal
    })

    external_results.append({
        "Model": name,
        **external
    })

internal_results_df = pd.DataFrame(internal_results)
external_results_df = pd.DataFrame(external_results)

print("\n================ INTERNAL CKD RESULTS ================")
display(internal_results_df.round(4))

print("\n================ EXTERNAL BD-KDD RESULTS ================")
display(external_results_df.round(4))

# ============================================================
# 6. CROSS-POPULATION COMPARISON
# ============================================================

comparison_df = internal_results_df.merge(
    external_results_df,
    on="Model",
    suffixes=("_Internal", "_External")
)

comparison_df["Accuracy_Gap"] = (
    comparison_df["Accuracy_Internal"]
    - comparison_df["Accuracy_External"]
)

comparison_df["F1_Gap"] = (
    comparison_df["F1_Internal"]
    - comparison_df["F1_External"]
)

comparison_df["AUC_Gap"] = (
    comparison_df["ROC_AUC_Internal"]
    - comparison_df["ROC_AUC_External"]
)

print("\n================ CROSS-POPULATION COMPARISON ================")
display(comparison_df.round(4))

# ============================================================
# 7. SELECT BEST MODEL USING EXTERNAL ROC-AUC
# ============================================================

best_model_name = (
    external_results_df
    .sort_values("ROC_AUC", ascending=False)
    .iloc[0]["Model"]
)

best_model = trained_models[best_model_name]

print("\nBEST EXTERNAL MODEL:", best_model_name)

# ============================================================
# 8. EXTERNAL CONFUSION MATRIX
# ============================================================

external_pred = best_model.predict(X_bd)
external_prob = best_model.predict_proba(X_bd)[:, 1]

cm = confusion_matrix(y_bd, external_pred)

print("\n================ EXTERNAL CLASSIFICATION REPORT ================")

print(
    classification_report(
        y_bd,
        external_pred,
        target_names=["Healthy", "Kidney Disease"],
        digits=4
    )
)

plt.figure(figsize=(6,5))
plt.imshow(cm)
plt.title("External BD-KDD Confusion Matrix - " + best_model_name)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks([0,1], ["Healthy", "Kidney Disease"])
plt.yticks([0,1], ["Healthy", "Kidney Disease"])

for i in range(2):
    for j in range(2):
        plt.text(j, i, cm[i,j], ha="center", va="center")

plt.colorbar()
plt.tight_layout()
plt.savefig("/content/Figure_1_External_Confusion_Matrix.png", dpi=300)
plt.show()

# ============================================================
# 9. ROC CURVE
# ============================================================

internal_prob = best_model.predict_proba(X_test)[:,1]

fpr_int, tpr_int, _ = roc_curve(y_test, internal_prob)
fpr_ext, tpr_ext, _ = roc_curve(y_bd, external_prob)

auc_int = roc_auc_score(y_test, internal_prob)
auc_ext = roc_auc_score(y_bd, external_prob)

plt.figure(figsize=(7,6))

plt.plot(
    fpr_int,
    tpr_int,
    label=f"Internal CKD AUC = {auc_int:.3f}"
)

plt.plot(
    fpr_ext,
    tpr_ext,
    label=f"External BD-KDD AUC = {auc_ext:.3f}"
)

plt.plot(
    [0,1],
    [0,1],
    linestyle="--",
    label="Random classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Internal vs External ROC - " + best_model_name)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("/content/Figure_2_Internal_External_ROC.png", dpi=300)
plt.show()

# ============================================================
# 10. PERMUTATION FEATURE IMPORTANCE
# ============================================================

perm = permutation_importance(
    best_model,
    X_bd,
    y_bd,
    n_repeats=30,
    random_state=42,
    scoring="roc_auc",
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "Feature": X_bd.columns,
    "Importance_Mean": perm.importances_mean,
    "Importance_SD": perm.importances_std
})

importance_df = (
    importance_df
    .sort_values("Importance_Mean", ascending=False)
    .reset_index(drop=True)
)

print("\n================ FEATURE IMPORTANCE ================")
display(importance_df.round(4))

plt.figure(figsize=(8,5))

plt.barh(
    importance_df["Feature"][::-1],
    importance_df["Importance_Mean"][::-1]
)

plt.xlabel("Mean decrease in ROC-AUC")
plt.ylabel("Clinical Feature")
plt.title("External Feature Importance - " + best_model_name)
plt.tight_layout()
plt.savefig("/content/Figure_3_Feature_Importance.png", dpi=300)
plt.show()

# ============================================================
# 11. AGE-GROUP SUBGROUP FAIRNESS ANALYSIS
# ============================================================

def age_fairness(model, X, y):

    d = X.copy()
    d["true"] = np.asarray(y)
    d["pred"] = model.predict(X)

    d["Age_Group"] = pd.cut(
        d["age"],
        bins=[0,40,60,200],
        labels=["<40","40-60",">60"],
        include_lowest=True
    )

    rows = []

    for group, g in d.groupby(
        "Age_Group",
        observed=True
    ):

        rows.append({
            "Age_Group": str(group),
            "N": len(g),
            "Accuracy": accuracy_score(
                g["true"],
                g["pred"]
            ),
            "Precision": precision_score(
                g["true"],
                g["pred"],
                zero_division=0
            ),
            "Recall": recall_score(
                g["true"],
                g["pred"],
                zero_division=0
            ),
            "F1": f1_score(
                g["true"],
                g["pred"],
                zero_division=0
            )
        })

    return pd.DataFrame(rows)

internal_fairness = age_fairness(
    best_model,
    X_test,
    y_test
)

external_fairness = age_fairness(
    best_model,
    X_bd,
    y_bd
)

print("\n================ INTERNAL AGE SUBGROUP ANALYSIS ================")
display(internal_fairness.round(4))

print("\n================ EXTERNAL AGE SUBGROUP ANALYSIS ================")
display(external_fairness.round(4))

# ============================================================
# 12. FAIRNESS GAPS
# ============================================================

recall_gap = (
    external_fairness["Recall"].max()
    - external_fairness["Recall"].min()
)

accuracy_gap = (
    external_fairness["Accuracy"].max()
    - external_fairness["Accuracy"].min()
)

f1_gap = (
    external_fairness["F1"].max()
    - external_fairness["F1"].min()
)

print("\nExternal Recall Gap:", round(recall_gap,4))
print("External Accuracy Gap:", round(accuracy_gap,4))
print("External F1 Gap:", round(f1_gap,4))

# ============================================================
# 13. FAIRNESS FIGURE
# ============================================================

external_fairness.set_index(
    "Age_Group"
)[
    ["Accuracy","Precision","Recall","F1"]
].plot(
    kind="bar",
    figsize=(9,6)
)

plt.title(
    "Age-Group Subgroup Performance - External BD-KDD"
)

plt.xlabel("Age Group")
plt.ylabel("Score")
plt.ylim(0,1)
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig(
    "/content/Figure_4_Age_Group_Fairness.png",
    dpi=300
)

plt.show()

# ============================================================
# 14. SAVE ALL IMPORTANT TABLES
# ============================================================

dataset_summary = pd.DataFrame({
    "Dataset": ["CKD","BD-KDD"],
    "Records": [len(X_ckd),len(X_bd)],
    "Features": [X_ckd.shape[1],X_bd.shape[1]],
    "Healthy": [
        int((y_ckd==0).sum()),
        int((y_bd==0).sum())
    ],
    "Kidney_Disease": [
        int((y_ckd==1).sum()),
        int((y_bd==1).sum())
    ]
})

output = "/content/FINAL_CKD_RESEARCH_RESULTS.xlsx"

with pd.ExcelWriter(output, engine="openpyxl") as writer:

    dataset_summary.to_excel(
        writer,
        sheet_name="Dataset_Summary",
        index=False
    )

    internal_results_df.to_excel(
        writer,
        sheet_name="Internal_Results",
        index=False
    )

    external_results_df.to_excel(
        writer,
        sheet_name="External_Results",
        index=False
    )

    comparison_df.to_excel(
        writer,
        sheet_name="Cross_Population",
        index=False
    )

    importance_df.to_excel(
        writer,
        sheet_name="Feature_Importance",
        index=False
    )

    internal_fairness.to_excel(
        writer,
        sheet_name="Internal_Age_Fairness",
        index=False
    )

    external_fairness.to_excel(
        writer,
        sheet_name="External_Age_Fairness",
        index=False
    )

print("\n================================================")
print("FINAL RESEARCH PIPELINE COMPLETED")
print("================================================")
print("Best model:", best_model_name)
print("Internal AUC:", round(auc_int,4))
print("External AUC:", round(auc_ext,4))
print("External Recall:", round(
    recall_score(y_bd, external_pred),4
))
print("External F1:", round(
    f1_score(y_bd, external_pred),4
))
print("External Accuracy:", round(
    accuracy_score(y_bd, external_pred),4
))
print("Age Recall Gap:", round(recall_gap,4))
print("Age Accuracy Gap:", round(accuracy_gap,4))
print("Age F1 Gap:", round(f1_gap,4))
print("\nExcel saved:")
print(output)
```
